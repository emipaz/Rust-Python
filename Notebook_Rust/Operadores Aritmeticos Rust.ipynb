# Operadores Aritméticos en Rust


Rust soporta los operadores aritméticos básicos de manera similar a la mayoría
de los lenguajes de programación. Sin embargo, tiene reglas estrictas sobre
**tipos de datos** que es importante conocer desde el principio.


## Operadores básicos



| Operador | Operación       | Ejemplo   |
|----------|-----------------|-----------|
| `+`      | Suma            | `5 + 3`   |
| `-`      | Resta           | `5 - 3`   |
| `*`      | Multiplicación  | `5 * 3`   |
| `/`      | División        | `5 / 3`   |
| `%`      | Módulo (resto)  | `5 % 3`   |


### 1 — Operadores con enteros

````rust
fn main() {
    let a = 10;
    let b = 3;

    println!("Suma:           {} + {} = {}", a, b, a + b);
    println!("Resta:          {} - {} = {}", a, b, a - b);
    println!("Multiplicación: {} * {} = {}", a, b, a * b);
    println!("División:       {} / {} = {}", a, b, a / b);   // trunca!
    println!("Módulo:         {} % {} = {}", a, b, a % b);
}
main()
````

### 2 — División entera

Cuando ambos operandos son enteros, Rust **trunca el resultado hacia cero**.

- `10 / 3` → `3`  (no `3.33...`)
- `-10 / 3` → `-3` (no `-4`, trunca hacia cero, no hacia abajo)

Para obtener un resultado decimal, los operandos deben ser de punto flotante.


### 2 — División y truncado

````rust
fn main() {
    // División entera — trunca
    println!("10 / 3   = {}", 10 / 3);     // 3
    println!("-10 / 3  = {}", -10 / 3);    // -3

    // División real — con f64
    println!("10.0 / 3.0 = {}", 10.0_f64 / 3.0_f64);  // 3.333...

    // Casteo al momento de dividir
    let x = 10;
    let y = 3;
    println!("cast: {}", x as f64 / y as f64);          // 3.333...
}
main()
````

### 3 — Tipos numéricos

Rust tiene varios tipos numéricos. Los más usados son:

**Enteros:**
- `i32` — entero con signo de 32 bits (el más común)
- `i64` — entero con signo de 64 bits
- `u32` — entero sin signo (solo positivos)
- `usize` — tamaño dependiente de la arquitectura (índices, longitudes)

**Punto flotante:**
- `f32` — flotante de 32 bits (menor precisión)
- `f64` — flotante de 64 bits ✅ (recomendado por defecto)

> Rust infiere el tipo automáticamente, pero siempre podés especificarlo.


````rust

fn main() {
    let entero: i32 = 100;
    let flotante: f64 = 3.14;
    let sin_signo: u32 = 42;

    println!("i32:  {}", entero);
    println!("f64:  {}", flotante);
    println!("u32:  {}", sin_signo);

    // Inferencia de tipo
    let inferido = 7;       // Rust infiere i32
    let inferido_f = 7.0;   // Rust infiere f64
    println!("inferido entero: {}", inferido);
    println!("inferido float:  {}", inferido_f);
}
main()
````

### 4  — Tipos estrictos

No se pueden mezclar tipos distintos en una operación aritmética directamente.
Es necesario convertir explícitamente usando `as`.


### Casteo con as

````rust
fn main() {
    let entero: i32 = 7;
    let flotante: f64 = 2.5;

    // Esto NO compila:
    // let resultado = entero + flotante;

    // Hay que castear:
    let resultado = entero as f64 + flotante;
    println!("7 + 2.5 = {}", resultado);  // 9.5

    // También se puede castear f64 → i32 (trunca los decimales)
    let truncado = flotante as i32;
    println!("2.5 as i32 = {}", truncado);  // 2
}
main()
````

### 5 — Potencia

Rust **no tiene operador `**`** como otros lenguajes.
En su lugar se usan métodos:

- `.pow(n)` — para enteros
- `.powi(n)` — para flotantes con exponente entero
- `.powf(n)` — para flotantes con exponente flotante
- `.sqrt()` — raíz cuadrada


### 5 — Potencia y raíz

````rust
fn main() {
    // Enteros
    let base: i32 = 2;
    println!("2^10 = {}", base.pow(10));       // 1024

    // Flotantes
    let x: f64 = 3.0;
    println!("3^4  = {}", x.powi(4));          // 81.0
    println!("2^0.5 = {}", 2.0_f64.powf(0.5)); // 1.4142... (√2)
    println!("√16  = {}", 16.0_f64.sqrt());    // 4.0
}
main()
````

### 6 — Más métodos matemáticos

Rust tiene muchos métodos disponibles directamente sobre los tipos `f64`:

```rust
f64::abs()    // valor absoluto
f64::floor()  // redondear hacia abajo
f64::ceil()   // redondear hacia arriba
f64::round()  // redondear al más cercano
f64::ln()     // logaritmo natural
f64::log2()   // logaritmo base 2
f64::log10()  // logaritmo base 10
f64::sin()    // seno (en radianes)
f64::cos()    // coseno
```


### 6 — Métodos matemáticos

````rust
fn main() {
    let n: f64 = -7.6;

    println!("abs:   {}", n.abs());      // 7.6
    println!("floor: {}", n.floor());    // -8.0
    println!("ceil:  {}", n.ceil());     // -7.0
    println!("round: {}", n.round());    // -8.0

    let m: f64 = 100.0;
    println!("log10: {}", m.log10());    // 2.0
    println!("ln:    {}", m.ln());       // 4.605...
    println!("sqrt:  {}", m.sqrt());     // 10.0

    // Constante PI incluida
    println!("PI:    {}", std::f64::consts::PI);
    println!("sin(PI/2): {}", (std::f64::consts::PI / 2.0).sin()); // 1.0
}
main()
````

## ⚠️ Overflow en Rust

Rust protege contra **desbordamiento de enteros**:

- En modo **debug** → el programa entra en pánico (`panic`) si hay overflow
- En modo **release** → hace wrapping silencioso (da vuelta al mínimo)

Para manejar esto explícitamente existen métodos especiales.


### 7 — Manejo de overflow

````rust
fn main() {
    let max = i32::MAX;
    println!("Máximo i32: {}", max);  // 2147483647

    // Wrapping: da vuelta sin pánico
    let wrapped = max.wrapping_add(1);
    println!("wrapping_add: {}", wrapped);  // -2147483648

    // Checked: devuelve Option, None si hay overflow
    let checked = max.checked_add(1);
    println!("checked_add: {:?}", checked);  // None

    // Saturating: se queda en el máximo/mínimo posible
    let saturated = max.saturating_add(1);
    println!("saturating_add: {}", saturated);  // 2147483647
}
main()
````

### 8 — Números complejos

A diferencia de algunos lenguajes, Rust **no tiene soporte nativo para
números complejos**. Sin embargo, existe el crate (librería) `num-complex`
que los implementa de forma completa.

Para usarlo en un proyecto real se agrega al `Cargo.toml`:
```toml
[dependencies]
num-complex = "0.4"
```


### 8 — Números complejos con num-complex


#### En Jupyter con el kernel de Rust se puede usar `:dep` para agregar dependencias.

````rust
:dep num-complex = "0.4"
use num_complex::Complex;

fn main() {
    let a = Complex::new(3.0_f64, 4.0);   // 3 + 4i
    let b = Complex::new(1.0_f64, -2.0);  // 1 - 2i

    println!("a = {}", a);           // 3+4i
    println!("b = {}", b);           // 1-2i
    println!("a + b = {}", a + b);   // 4+2i
    println!("a * b = {}", a * b);   // 11-2i
    println!("|a| = {}", a.norm());  // 5.0  (módulo)
    println!("a² = {}", a.powi(2));  // -7+24i
}
main()
````

````rust
:dep num-complex = "0.4"
use num_complex::Complex;

fn main() {
    // Representás el -2 como un complejo con parte imaginaria 0
    let x = Complex::new(-2.0_f64, 0.0);
    let resultado = x.sqrt();
    println!("√(-2) = {}", resultado);  // 0+1.4142135623730951i
}
main()
````

### Resumen



| Concepto                  | Comportamiento en Rust                          |
|---------------------------|-------------------------------------------------|
| División entera           | Trunca hacia cero                               |
| Mezcla de tipos           | No permitida, hay que usar `as`                 |
| Potenciación              | `.pow()`, `.powi()`, `.powf()`                  |
| Overflow                  | Pánico en debug, wrapping en release            |
| Números complejos nativos | ❌ — se usa el crate `num-complex`              |
| Funciones matemáticas     | Disponibles como métodos sobre `f64`            |

## Notación Científica



Rust soporta notación científica de forma nativa usando `e` o `E` como separador
del exponente, de la misma forma que la mayoría de los lenguajes modernos.

La sintaxis es: `<mantisa>e<exponente>`
```rust
3.2e-23   // 3.2 × 10⁻²³
6.022e23  // 6.022 × 10²³
1.5E10    // 1.5 × 10¹⁰
```

> El tipo recomendado para estos valores es siempre `f64` ya que
> `f32` puede perder precisión con exponentes muy grandes o muy pequeños.


### 1 — Declaración y operaciones

````rust
fn main() {
    let masa_electron: f64 = 9.109e-31;    // kg
    let avogadro:      f64 = 6.022e23;     // mol⁻¹
    let velocidad_luz: f64 = 2.998e8;      // m/s

    println!("Masa del electrón:  {:e}", masa_electron);
    println!("Número de Avogadro: {:e}", avogadro);
    println!("Velocidad de la luz:{:e}", velocidad_luz);

    // Operaciones entre notaciones científicas
    let energia = masa_electron * velocidad_luz.powi(2);
    println!("\nE = mc²");
    println!("E = {:e} J", energia);
}
main()
````

### Formato de salida



Rust ofrece varias formas de mostrar números en notación científica
usando especificadores de formato dentro de `println!`:

| Formato   | Descripción                          | Ejemplo         |
|-----------|--------------------------------------|-----------------|
| `{:e}`    | Notación científica con `e` minúscula| `6.022e23`      |
| `{:E}`    | Notación científica con `E` mayúscula| `6.022E23`      |
| `{:.2e}`  | Controla cantidad de decimales       | `6.02e23`       |
| `{:.5e}`  | Más precisión en la mantisa          | `6.02214e23`    |
| `{}`      | Formato estándar (puede ser largo)   | `602200000000...`|


### 2 — Formatos de salida


````rust
fn main() {
    let avogadro: f64 = 6.02214076e23;

    println!("Estándar:       {}",    avogadro);  // número completo
    println!("Científica:     {:e}",  avogadro);  // 6.02214076e23
    println!("Mayúscula:      {:E}",  avogadro);  // 6.02214076E23
    println!("2 decimales:    {:.2e}", avogadro); // 6.02e23
    println!("5 decimales:    {:.5e}", avogadro); // 6.02214e23

    // También funciona con negativos
    let constante_planck: f64 = 6.626e-34;
    println!("\nPlanck:         {:e}",   constante_planck);
    println!("Planck (3 dec): {:.3e}", constante_planck); // 6.626e-34
}
main()
````

### 3 — Límites y precisión


`f64` puede representar números en el rango aproximado de `10⁻³⁰⁸` a `10³⁰⁸`.
Rust expone estas constantes directamente:
```rust
f64::MAX        // ~1.8 × 10³⁰⁸
f64::MIN        // ~-1.8 × 10³⁰⁸
f64::MIN_POSITIVE // ~2.2 × 10⁻³⁰⁸  (positivo más pequeño)
f64::INFINITY   // ∞
f64::NAN        // Not a Number
```

Si el resultado de una operación supera estos límites, Rust devuelve
`inf` en lugar de producir un error.


### 3 — Límites y casos extremos


````rust
fn main() {
    println!("Máximo f64:   {:e}", f64::MAX);           // ~1.8e308
    println!("Mínimo pos:   {:e}", f64::MIN_POSITIVE);  // ~2.2e-308

    // Qué pasa al superar los límites
    let overflow = f64::MAX * 10.0;
    println!("MAX * 10:     {}", overflow);  // inf

    let underflow = f64::MIN_POSITIVE / 1e200;
    println!("MIN / 1e200:  {:e}", underflow);  // 0e0

    // Detectar estos casos
    println!("¿Es infinito? {}", overflow.is_infinite());  // true
    println!("¿Es NaN?      {}", f64::NAN.is_nan());       // true
    println!("¿Es finito?   {}", (3.2e10_f64).is_finite()); // true
}
main()
````
