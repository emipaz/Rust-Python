# Funciones en Rust a Fondo



Las funciones en Rust son pan de cada día (`fn`). Ya hemos estado usando la función `main`, que es el punto de entrada de todo programa Rust.

En este cuaderno profundizaremos en:
- Cómo definir parámetros y argumentos.
- Cómo retornar valores.
- **La gran diferencia en Rust: Sentencias (Statements) vs Expresiones (Expressions).**


## 1. Declaración de una Función y Parámetros



En Rust, el estilo de nombrado (convención) para funciones y variables es `snake_case` (todas las letras en minúscula y las palabras separadas por guion bajo). 

Además, en las firmas de las funciones, **debes declarar explícitamente el tipo de cada parámetro**. Esto es innegociable y ayuda al compilador de Rust a no tener que adivinar tipos en todo tu código.

````rust
fn saludar_persona(nombre: &str, edad: i32) {
    println!("¡Hola {}! Tienes {} años.", nombre, edad);
}

fn main() {
    saludar_persona("Carlos", 28);
    saludar_persona("Ana", 32);
}

main()
````

## 2. El corazón de Rust: Sentencias vs Expresiones

Para entender cómo las funciones devuelven datos en Rust, primero debes entender esta regla de oro:

- **Sentencias (Statements):** Son instrucciones que realizan una acción y **no** devuelven un valor. (Terminan en `;`).
- **Expresiones (Expressions):** Evalúan a un valor y **sí** lo devuelven.

Crear una variable (`let y = 6;`) es una *Sentencia*. No devuelve nada. Por lo tanto, no puedes hacer algo como `let x = (let y = 6);`.
Un bloque de código envuelto en llaves `{}` es una *Expresión*, y su valor de retorno es siempre **la última línea del bloque si esta no tiene punto y coma (`;`)**.

````rust
fn main() {
    let y = {
        let x = 3;
        // Esta línea no tiene punto y coma (;), 
        // por lo tanto, toda esta llave {...} "devuelve" (retorna) x + 1
        x + 1
    };

    println!("El valor de 'y' es: {}", y);
}

main()
````

## 3. Retornando Valores de una Función


Para indicar que una función devuelve un valor, usamos una flecha `->` seguida del tipo de dato.
Tienes dos formas de retornar el valor:

1. **Implícito (El método idiomático de Rust):** Dejando la última expresión de la función **sin punto y coma**.
2. **Explícito (Early return):** Usando la palabra clave `return` acompañada de un `;` (normalmente usado si queremos salir de la función antes de llegar al final).

````rust
// Ejemplo Idiomático (Retorno implícito)
fn sumar_cinco(x: i32) -> i32 {
    x + 5 // ¡OJO! No lleva punto y coma al final
}

// Ejemplo con Return explícito
fn restar_cinco(x: i32) -> i32 {
    return x - 5; // Aquí sí lleva punto y coma porque es una Statement
}

fn main() {
    let a = sumar_cinco(10);
    let b = restar_cinco(10);
    
    println!("10 + 5 = {}", a);
    println!("10 - 5 = {}", b);
}

main()
````

## 4. Retorno "Unit" `()`

Si una función no tiene una flecha de retorno `->`, entonces implícitamente devuelve el tipo Unit (Unidad), que se representa como `()`. 

Es equivalente a decir que la función "no devuelve nada útil". Es parecido al `void` en C o Java.

````rust
// Estas dos funciones de abajo son exactamente iguales para Rust

fn funcion_vacia_implicita() {
    println!("Devuelvo Unit implicitamente.");
}

fn funcion_vacia_explicita() -> () {
    println!("Devuelvo Unit explicitamente.");
}

fn main() {
    let res1 = funcion_vacia_implicita();
    let res2 = funcion_vacia_explicita();
    
    // Podemos demostrar que ambas devuelven un valor vacío `()`
    println!("Resultado 1 = {:?}", res1);
    println!("Resultado 2 = {:?}", res2);
}

main()
````

## 📌 Cosas a recordar

1. Firma estricta: Siempre debes definir de qué tipo son los parámetros (`n: i32`) y de qué tipo es el retorno (`-> i32`).
2. Si omites el `;` en la última línea matemática o lógica de una función, ese se convierte en el valor devuelto de la función.
3. Si pones un `;` al final por accidente (`x + 5;`), esto se convierte en una *Sentencia* que devolverá `()`. Si Rust esperaba un `i32`, ¡el compilador te dará un error rojo gigante!

## 5. El manejo de parámetros opcionales


En Rust, las firmas de las funciones son un contrato inquebrantable. A diferencia de lenguajes más permisivos, **no existen parámetros omitibles por defecto**. El compilador de Rust siempre exigirá que envíes exactamente el número y tipo de argumentos declarados en la función.

Entonces, ¿cómo manejamos situaciones donde un argumento es opcional o puede no estar disponible? Usando el superpoder de Rust: **el enumerador `Option<T>`**.

El tipo `Option` obliga estricta y visualmente al programador a declarar si está enviando un dato válido (`Some(valor)`) o si no está enviando nada (`None`).

````rust
// El parámetro 'saludo' espera un texto protegido por Option (puede tener algo o estar vacío)
fn saludar(nombre: &str, saludo: Option<&str>) {
    
    // El método `.unwrap_or()` nos permite abrir de forma segura el 'Option'.
    // Si viene con un 'None', usa la palabra "Hola" como valor de rescate (nuestro default real).
    let saludo_final = saludo.unwrap_or("Hola");
    
    println!("{}, {}!", saludo_final, nombre);
}

fn main() {
    // 1. Llamada clásica sin un saludo específico (pasamos explícitamente None)
    saludar("Carlos", None);


    // 2. Llamada donde sí queremos modificar el saludo (pasamos Some)
    saludar("Ana", Some("Bienvenida"));

}
main()
````

> **Nota**

En lenguajes flexibles a veces puedes omitir un parámetro y que este tome un valor por defecto automáticamente. 

**En Rust, esto es ilegal**. La firma de las funciones es un contrato inquebrantable y el compilador siempre exigirá que envíes exactamente el número declarado de argumentos. **Nunca puedes omitir un argumento al llamar la función**.

Entonces, ¿qué hacemos si funcionalmente un dato es "opcional"? Usamos el enumerador **`Option<T>`**.
Al usar `Option`, la función sigue pidiendo el argumento obligatoriamente, pero a cambio te permite pasar la palabra **`None`** ("aquí no hay nada") o **`Some(valor)`** ("aquí tienes un valor"). 

Esto hace que el código sea extremadamente seguro y fácil de leer, porque cualquiera que vea tu llamada a la función sabrá que omitiste ese dato explícitamente y no por accidente.

## 6. Funciones y Parámetros Estructurados

A medida que desarrolles, inevitablemente crearás "funciones gigantes" que requieren demasiados parámetros de decisión o configuración:

```rust
fn iniciar(host: &str, puerto: u16, log_level: &str, es_seguro: bool)`
```

La forma totalmente **idiomática en Rust** para no construir parámetros kilométricos es agruparlos en un `Struct`.
Pero aquí entra una palabra clave nueva: **`impl`** (Implementación). 

En Rust, la palabra `struct` solo define la "forma" de los datos, pero no hace nada por sí sola. 

Para agregar comportamientos, métodos o lógicas a esa estructura, abrimos un bloque `impl`. 

En este caso puntual, usaremos `impl` para enseñarle a Rust cómo generar una versión de "fábrica" o predeterminada de nuestro `Struct` mediante el rasgo estándar llamado `Default`.

Si combinamos un `Struct` con el trait (rasgo) de Rust llamado **`Default`**, logramos crear llamadas a funciones súper limpias donde el usuario visualiza qué campo está configurando por el nombre, pudiendo ignorar el resto.


````rust
// 1. Creamos el esqueleto de nuestros datos
#[derive(Debug)] 

struct Configuracion {
    host: String,
    puerto: u16,
    log_level: String,
}

// 2. Abrimos un bloque 'impl' (Implementación). 
// Aquí le "implementamos" el rasgo Default a nuestra Estrutura.
// Es decir, le enseñamos al compilador cuáles son los valores de fábrica.
impl Default for Configuracion {
    fn default() -> Self {
        Configuracion {
            host: String::from("localhost"),
            puerto: 8080,
            log_level: String::from("INFO"),
        }
    }
}

// 3. Nuestra función principal ahora toma solo UN argumento (el struct)
fn arrancar_aplicacion(conf: Configuracion) {
    println!("Arrancando la aplicación usando:\n{:#?}", conf);
}

fn main() {
    // Caso A: Usar absolutamente todos los valores del bloque impl Default
    let conf_estandar = Configuracion::default();
    arrancar_aplicacion(conf_estandar);

    println!("-------------------");

    // Caso B: Sobrescribir el puerto, y pedirle que rellene el resto
    // usando los valores que declaramos en el bloque impl Default
    let conf_avanzada = Configuracion {
        puerto: 3000, 
        ..Configuracion::default() 
    };
    
    arrancar_aplicacion(conf_avanzada);
}
main()

````

**Detalle Técnico Importante: ¿Qué es `Self` con mayúscula?**

En el código anterior usamos: `fn default() -> Self { ... }`

Si vienes de otros lenguajes, podrías pensar que `Self` es solo una convención que puedes renombrar a gusto. **En Rust, `Self` (con 'S' mayúscula) es una palabra clave estricta y reservada**. 

Funciona como un "alias dinámico" que significa literalmente: *"el tipo de dato exacto que estamos implementando en este bloque"*. 
En nuestro ejemplo, escribir `Self` es **exactamente idéntico** a escribir `Configuracion`. Podríamos haber escrito:

`fn default() -> Configuracion {` Y el código funcionaría perfecto. 

Sin embargo, usar `Self` es la norma (idiomático en Rust), ya que si mañana decides renombrar tu `Struct` a `ConfiguracionServer`, no tendrás que cambiar a mano el tipo de retorno en todos tus métodos internos; `Self` se actualizará automáticament

## 7. Argumentos de longitud ilimitada

En Rust, las funciones técnicas regulares no pueden recibir una "cantidad arbitraria e indefinida de elementos sueltos" (variadic arguments). Su firma debe ser invariable y rígida por seguridad de memoria.

Si tu función necesita procesar "1, 3, o 100" ítems abstractos de un mismo tipo de dato, la solución es exigir un agrupador. El argumento ideal para exigir es un **Slice** (`&[T]`), ya que significa *"dame una vista/referencia hacia un grupo de datos, sin importar si lo tienes en un Array o en un Vector"*.

````rust
// 'numeros' es un slice (un grupo arbitrario de elementos i32 accesibles en fila)
fn calcular_suma_total(numeros: &[i32]) -> i32 {
    let mut acumulado = 0;
    for valor in numeros {
        acumulado += valor;
    }
    acumulado
}

fn main() {
    // Variante 1: Pasando un slice hacia un Vector pre-existente
    let mis_ganancias = vec![100, 250, 50, 10];
    let total = calcular_suma_total(&mis_ganancias);
    println!("El total del vector es: {}", total);
    
    // Variante 2: Pasando un slice hacia un Array literal introducido 'al vuelo'
    let total2 = calcular_suma_total(&[5, 5, 5]);
    println!("El total del array estático es: {}", total2);
}
main()

// Nota técnica:
// Si realmente te preguntas cómo funciones como `println!` logran aceptar tantos 
// términos variables separados por comas, es porque estrictamente no son "funciones",
// sino que son "Macros" (llevan el sufijo !). Las macros operan bajo otras reglas
// directamente expandiendo y alterando el código fuente antes de la fase de compilación.
````

## 8. Cuidado con el Secuestro de Datos (Ownership en Parámetros)

En Rust, cuando pasas un dato simple (como un número `i32` o un `bool`) a una función, Rust hace una copia automática porque son baratos. 

Pero cuando le pasas un dato complejo (como un `String` o un `Vector`), **la función toma "propiedad" total de ese dato**. 

Cuando la función termina, el dato se destruye y ya no existe para el resto del programa. ¡Rust te lanzará un error si intentas usarlo después de dárselo a la función!

````rust
fn procesar_texto(texto: String) {
    println!("Procesando: {}", texto);
} // <- Aquí la variable 'texto' muere para siempre (memoria liberada)

fn main() {
    let mi_mensaje = String::from("Hola Rust");
    
    procesar_texto(mi_mensaje);
    
    // ❌ DESCOMENTA LA LÍNEA DE ABAJO Y MIRA EL ERROR ROJO:
    println!("Quiero usar mi mensaje de nuevo: {}", mi_mensaje);
}
main()
````

### 8.1 La Solución: Préstamos (`&` References)

Tú casi nunca quieres perder tus variables tras pasarlas a una función. 

La solución es **no regalar la variable, sino "prestarla"**.

Para prestar un dato, usamos el símbolo de referencia (`&`), lo que en Rust se llama un `Borrowing`. 

El dueño original de la variable la conserva, y la función solo recibe "permiso para mirarla" temporalmente.

````rust
// Fíjate en el '&' -> La función ahora pide 'prestado' el String
fn leer_texto(texto_prestado: &String) {
    println!("Leyendo prestado: {}", texto_prestado);
} // <- Aquí se devuelve el préstamo, el dato NO se destruye

fn main() {
    let mi_documento = String::from("Reporte Clasificado");
    
    // Le pasamos la referencia &mi_documento (no el dueño original)
    leer_texto(&mi_documento);
    
    // ✅ ¡Felicidades! La variable sigue existiendo y podemos seguir usándola
    println!("Sigo siendo dueño de: {}", mi_documento);
}
main()

// Tip Extra: En Rust idiomático, en lugar de pedir `&String`, las funciones 
// suelen pedir directamente `&str` (un slice de texto), porque es aún más flexible.
````
