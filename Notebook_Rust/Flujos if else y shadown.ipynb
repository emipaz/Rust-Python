# Condicionales en Rust



En Rust, las estructuras condicionales controlan el flujo de ejecución del programa.
A diferencia de otros lenguajes, en Rust **`if` es una expresión**, lo que significa
que puede devolver un valor.


## 1 — if básico

```rust
fn main() {
    let temperatura = 30;

    if temperatura > 25 {
        println!("Hace calor ☀️");
    } else {
        println!("Está fresco 🌧️");
    }
}
main()
```

## 2 — if / else if / else


Podemos encadenar múltiples condiciones usando `else if`.


```rust
fn main() {
    let nota = 75;

    if nota >= 90 {
        println!("Sobresaliente");
    } else if nota >= 70 {
        println!("Aprobado");
    } else if nota >= 50 {
        println!("Regular");
    } else {
        println!("Reprobado");
    }
}
main()
```

## 3 — if como expresión



En Rust, `if` puede usarse para **asignar un valor** a una variable.
Ambas ramas deben devolver el **mismo tipo**.


```rust
fn main() {
    let es_mayor = true;

    let mensaje = if es_mayor {
        "Puede ingresar ✅"
    } else {
        "Acceso denegado ❌"
    };

    println!("{}", mensaje);
}
main()
```

### 3.1 Condicional con Opcion

```rust
fn print_number(number: Option<i32>) {
    if let Some(n) = number {
        println!("The number is {}", n);
    } else {
        println!("There is no number");
    }
}

fn main() {
    print_number(Some(42));
    print_number(None);

    let num = 12;
    print_number(Some(num));
}
main()
```

## 4 — loop



`loop` ejecuta un bloque indefinidamente hasta encontrar un `break`.
También puede **devolver un valor** a través del `break`.


### 4 — loop con valor de retorno


```rust
fn main() {
    let mut contador = 0;

    let resultado = loop {
        contador += 1;
        if contador == 5 {
            break contador * 2; // devuelve 10
        }
    };

    println!("El resultado del loop es: {}", resultado);
}
main()
```

## 5 — while


Ejecuta el bloque **mientras** una condición sea verdadera.


```rust
fn main() {
    let mut n = 1;

    while n <= 5 {
        println!("n = {}", n);
        n += 1;
    }
}
main()
```

## 6 — for


`for` es la forma más idiomática de iterar en Rust.
Se usa con rangos o colecciones.


### 6 — for con rango


```rust
fn main() {
    // Rango exclusivo: 1..5 → 1, 2, 3, 4
    for i in 1..5 {
        println!("i = {}", i);
    }

    println!("---");

    // Rango inclusivo: 1..=5 → 1, 2, 3, 4, 5
    for i in 1..=5 {
        println!("i = {}", i);
    }
}
main()
```

```rust
fn tabla_de_multiplicar(n: u32) {
    // Imprime la tabla de multiplicar del número n del 1 al 10
    // el raqngo es inclusivo para incluir el 10
    for i in 1..11 { // se puede usar 1..=10
        println!("{} x {} = {}", n, i, n * i);
    }
}
tabla_de_multiplicar(7);
```

## 7 — match


`match` compara un valor contra varios **patrones**.
Es exhaustivo: Rust obliga a cubrir **todos los casos posibles**.


```rust
fn main() {
    let dia = 3;

    let nombre = match dia {
        1 => "Lunes",
        2 => "Martes",
        3 => "Miércoles",
        4 => "Jueves",
        5 => "Viernes",
        6 | 7 => "Fin de semana",  // múltiples patrones con |
        _ => "Día inválido",        // comodín obligatorio
    };

    println!("Hoy es {}", nombre);
}
main()
```

## 8 — match con rangos y guards


```rust

fn main() {
    let puntaje = 82;

    let calificacion = match puntaje {
        90..=100 => "A",
        80..=89  => "B",
        70..=79  => "C",
        60..=69  => "D",
        _        => "F",
    };

    println!("Calificación: {}", calificacion);
}
main()
```

# 2. Shadowing de Variables en Rust

## 8 — Introducción al Shadowing



**Shadowing** (sombreado) ocurre cuando declaramos una nueva variable
con el **mismo nombre** que una anterior usando `let`.

La nueva variable *oculta* a la anterior dentro de su scope.
Esto es diferente a `mut`: con shadowing podemos incluso **cambiar el tipo**.


## 9 — Shadowing básico


```rust
fn main() {
    let x = 5;
    println!("x original: {}", x);

    let x = x + 10; // nueva variable x que sombrea a la anterior
    println!("x después del shadow: {}", x);

    let x = x / 2;
    println!("x después del segundo shadow: {}", x);

}
main()
```

## 9.1 — Shadowing vs mut


### Shadowing vs `mut`

| Característica         | `mut`            | Shadowing (`let`)       |
|------------------------|------------------|--------------------------|
| Cambia el valor        | ✅               | ✅                       |
| Cambia el tipo         | ❌               | ✅                       |
| La variable es inmutable después | ✅ (por defecto) | ✅             |
| Requiere `let` de nuevo | ❌              | ✅                       |


## 10 — Shadowing cambiando el tipo


```rust
fn main() {
    // Primero es un &str (texto)
    let dato = "42";
    println!("dato como texto: {}", dato);

    // Ahora es un i32 (entero) — imposible con mut
    let dato = dato.parse::<i32>().unwrap();
    println!("dato como número: {}", dato + 8);
}
main()
```

## 11 — Shadowing en bloques (scope)


```rust
fn main() {
    let color = "azul";

    {
        let color = "rojo"; // shadow solo dentro de este bloque
        println!("Dentro del bloque: {}", color); // rojo
    }

    println!("Fuera del bloque: {}", color); // azul — el original sigue intacto
}
main()
```


### 10 — Cuándo usar shadowing


- ✅ Al transformar un valor y no necesitar el nombre anterior
- ✅ Al parsear o convertir tipos (texto → número, etc.)
- ✅ Para evitar nombres redundantes como `texto`, `texto_parseado`, `texto_final`
- ⚠️ Usar con cuidado: abusar del shadowing puede hacer el código difícil de leer
